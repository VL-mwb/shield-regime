## 📈 Step 2: Fetch Financial Market Data

We will use `yfinance` to fetch the historical daily data of GameStop ($GME$) from late 2020 through early 2021. This provides a perfect, real-world retail-driven price kinematics case study.

```python
import yfinance as yf
import pandas as pd

# Download historical daily GME data
print("📥 Fetching GameStop (GME) stock data...")
df_raw = yf.download("GME", start="2020-11-01", end="2021-04-01")

# Format columns and index to match ShieldRegime requirements
df_gme = df_raw.copy()

# Handle MultiIndex columns (common in newer yfinance versions)
if isinstance(df_gme.columns, pd.MultiIndex):
    df_gme.columns = [col[0].lower() for col in df_gme.columns]
else:
    df_gme.columns = [col.lower() for col in df_gme.columns]

df_gme.index = pd.to_datetime(df_gme.index)

print(f"✅ Successfully loaded {len(df_gme)} trading sessions of GME data!")
df_gme.head()
```

## 🧠 Step 3: Initialize and Run the Pump & Dump Detector

We will run the `PumpDumpDetector` using the correct `scan()` method from the API to compute smooth velocity, acceleration, volume profiles, and trigger structural regime state-machine alerts.

```python
from shield_regime.detectors.pump_dump import PumpDumpDetector

# Initialize the detector with an optimized lookback window of 40 days
# This ensures proper baseline calculation for the explosive GME short squeeze
detector = PumpDumpDetector(
    velocity_z_threshold=2.5,
    volume_spike_threshold=3.0,
    lookback=40
)

# Run detection using the official scan() API
alerts = detector.scan(df_gme, ticker="GME")

print(f"🚨 Detected {len(alerts)} manipulation regime triggers!")
for i, alert in enumerate(alerts, 1):
    print(f"  [{i}] [{alert.severity.value}] {alert.alert_type}")
    print(f"       Window: {alert.start_date.strftime('%Y-%m-%d')} → {alert.end_date.strftime('%Y-%m-%d')} ({alert.duration_days} days)")
    print(f"       Headline: {alert.headline}")
```

## 🎨 Step 4: Render the Interactive Kinematics Visualization

Now, we will generate the stunning interactive Plotly chart using the official `plot_alerts` visualization API.

The chart shows:
- 🔴 **Red shaded zone** = detected manipulation window
- 🟢 **Green triangle** = pump start date
- 🔻 **Red triangle** = dump low date
- 📊 **Bottom panel** = Price Velocity Z-Score (exceeding ±2.5σ triggers an alert)

```python
from shield_regime.visualizer import plot_alerts

# Generate the interactive chart using plot_alerts()
fig = plot_alerts(df_gme, alerts, ticker="GME", lookback=40)

# Display the figure in your browser/Colab cell
if fig is not None:
    fig.show()
```
